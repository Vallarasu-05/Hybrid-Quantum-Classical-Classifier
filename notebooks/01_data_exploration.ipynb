```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from src.data_preprocessing import load_binary_iris, TASKS

iris = load_iris(as_frame=True)
df = iris.frame
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, (task, (a, b)) in zip(axes, TASKS.items()):
    sub = df[df.target.isin([a, b])]
    for cls, col in zip((a, b), ("tab:blue", "tab:red")):
        s = sub[sub.target == cls]
        ax.scatter(s["petal length (cm)"], s["petal width (cm)"], c=col, label=iris.target_names[cls], alpha=.7)
    ax.set(title=task.replace("_", " "), xlabel="petal length (cm)", ylabel="petal width (cm)"); ax.legend()
plt.tight_layout(); plt.show()
```

```python
d = load_binary_iris("versicolor_vs_virginica", seed=0)
print("train:", tuple(d.X_train.shape), " test:", tuple(d.X_test.shape))
print("class balance train/test:", int(d.y_train.sum()), "/", len(d.y_train), "|", int(d.y_test.sum()), "/", len(d.y_test))
print("train mean (should be ~0):", d.X_train.mean(0).numpy().round(3))
print("test  mean (NOT forced to 0 -> no leakage):", d.X_test.mean(0).numpy().round(3))
```
